#### 1.导入基础包

```python
from utils.dataLoad import load_YiChang_with_classes
from sklearn.model_selection import train_test_split
from imblearn.metrics import classification_report_imbalanced
from sklearn.metrics import classification_report
import pandas as pd
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)

from models.RacAdvancedClassifier import RacAdvancedClassifier
```

```python
def make_a_test():
    rac = RacAdvancedClassifier(kernel_type='rbf', kernel_pars=[10.7], c=7.5)
    rac.fit(x_train_res,y_train_res)
    y_pred_train = rac.predict(x_train_res)
    y_pred_test = rac.predict(x_test)

    print("RAC train predict Results:\n", classification_report(y_train_res, y_pred_train, digits=4))
    print("RAC test predict Results:\n", classification_report(y_test, y_pred_test, digits=4))
```

#### 2.加载数据

```python
data = load_YiChang_with_classes()

# 将数据分为标签和特征，并转numpy数组
y = data.iloc[:, 0].to_numpy()
X = data.iloc[:, 1:].to_numpy()

```

#### 3.划分数据集

```python
# 数据划分，70%训练集，30%测试集
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42,stratify=y)
```

#### 4.重采样

```python
from imblearn.under_sampling import RandomUnderSampler, NearMiss, CondensedNearestNeighbour, EditedNearestNeighbours, RepeatedEditedNearestNeighbours, AllKNN, InstanceHardnessThreshold, NeighbourhoodCleaningRule
from imblearn.over_sampling import RandomOverSampler, SMOTE, ADASYN, BorderlineSMOTE, SVMSMOTE, SMOTENC, KMeansSMOTE
from utils.resample.CTGAN import CTGAN
from imblearn.combine import SMOTETomek, SMOTEENN
```

##### baseline

```python
x_train_res, y_train_res = x_train, y_train
make_a_test()
```

##### 欠采样

```python
rus = RandomUnderSampler(random_state=42, sampling_strategy='auto')
x_train_res, y_train_res = rus.fit_resample(x_train, y_train)
make_a_test()
```

```python
near_miss = NearMiss(sampling_strategy='auto')
x_train_res, y_train_res = near_miss.fit_resample(x_train, y_train)
make_a_test()
```

##### 过采样

```python
ros = RandomOverSampler(random_state=42, sampling_strategy='auto')
x_train_res, y_train_res = ros.fit_resample(x_train, y_train)
make_a_test()
```

```python
smote = SMOTE(random_state=42, sampling_strategy='auto')
x_train_res, y_train_res = smote.fit_resample(x_train, y_train)
make_a_test()
```

```python
adasyn = ADASYN(random_state=42, sampling_strategy='auto')
x_train_res, y_train_res = adasyn.fit_resample(x_train, y_train)
make_a_test()
```

```python
borderline_smote = BorderlineSMOTE(random_state=42, sampling_strategy='auto')
x_train_res, y_train_res = borderline_smote.fit_resample(x_train, y_train)
make_a_test()
```

```python
svm_smote = SVMSMOTE(random_state=42, sampling_strategy='auto')
x_train_res, y_train_res = svm_smote.fit_resample(x_train, y_train)
make_a_test()
```

```python
smoteNc = SMOTENC(categorical_features=[0, 3], random_state=42, sampling_strategy='auto')
x_train_res, y_train_res = smoteNc.fit_resample(x_train, y_train)
make_a_test()
```

```python
kmeans_smote = KMeansSMOTE(random_state=42,sampling_strategy='auto',cluster_balance_threshold=0.05)
x_train_res, y_train_res = kmeans_smote.fit_resample(x_train, y_train)
make_a_test()
```

```python
ctgan = CTGAN()
x_train_res, y_train_res = ctgan.fit_resample(x_train, y_train)
make_a_test()
```

##### 混合重采样

```python
smote_tomek = SMOTETomek(random_state=42, sampling_strategy='auto')
x_train_res, y_train_res = smote_tomek.fit_resample(x_train, y_train)
make_a_test()
```

```python
smote_enn = SMOTEENN(random_state=42, sampling_strategy='auto')
x_train_res, y_train_res = smote_enn.fit_resample(x_train, y_train)
make_a_test()
```
